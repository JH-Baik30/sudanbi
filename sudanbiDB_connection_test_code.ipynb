이후 접속 도커 compose down & up 으로 정상 접속 확인 함

````python
import mysql.connector
from mysql.connector import Error

MYSQL = {
    "host": "172.30.1.23",
    "port": 3306,
    "user": "webUser",
    "password": "12344321",
    "database": "sudanbi_DB",
}

def test_connection():
    try:
        conn = mysql.connector.connect(**MYSQL)

        if conn.is_connected():
            cursor = conn.cursor()

            cursor.execute("SELECT VERSION()")
            version = cursor.fetchone()
            print(f"✅ DB 연결 성공!")
            print(f"   MySQL 버전: {version[0]}")

            cursor.execute("SHOW TABLES")
            tables = cursor.fetchall()
            print(f"   테이블 목록:")
            for table in tables:
                print(f"     - {table[0]}")

            cursor.execute("SELECT COUNT(*) FROM question_item")
            count = cursor.fetchone()
            print(f"   question_item 레코드 수: {count[0]}")

            cursor.close()
            conn.close()

    except Error as e:
        print(f"❌ 연결 실패: {e}")

if __name__ == "__main__":
    test_connection()
````
